```python
import json
from itertools import islice
from pathlib import Path
import pandas as pd
```

```python
file = Path("9.listens")

print("File found:", file.exists())
print("File size (MB):", round(file.stat().st_size / 1_000_000, 2))
```

```python
with file.open(encoding="utf-8") as f:
    records = [json.loads(line) for line in islice(f, 10000)]

data = pd.json_normalize(records)

print(data.shape)
data.head()
```

```python
print("Shape:", data.shape)

print("\nColumns:")
print(data.columns.tolist())

print("\nData types:")
print(data.dtypes)
```

```python
relevant_cols = [
    "user_id",
    "user_name",
    "timestamp",
    "recording_msid",
    "track_metadata.track_name",
    "track_metadata.artist_name",
    "track_metadata.release_name"
]

data[relevant_cols].head()
```

```python
data[relevant_cols].isna().sum()
```

```python
clean = data.copy()
```

```python
clean["timestamp_clean"] = pd.to_datetime(
    clean["timestamp"],
    unit="s",
    utc=True,
    errors="coerce"
)

print("Failed timestamp conversions:", clean["timestamp_clean"].isna().sum())

clean[["timestamp", "timestamp_clean"]].head()
```

```python
text_cols = [
    "user_name",
    "track_metadata.track_name",
    "track_metadata.artist_name",
    "track_metadata.release_name"
]

for col in text_cols:
    changed = (
        clean[col].notna() &
        (clean[col] != clean[col].str.strip())
    ).sum()
    
    print(col, ":", changed)
```

```python
for col in text_cols:
    clean[col] = clean[col].str.strip()

for col in text_cols:
    changed = (
        data[col].notna() &
        clean[col].notna() &
        (data[col] != clean[col])
    ).sum()
    
    print(col, "changed:", changed)
```

```python
exact_duplicates = pd.Series(
    [json.dumps(record, sort_keys=True) for record in records]
).duplicated().sum()

repeated_recording_ids = data["recording_msid"].duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)
print("Repeated recording_msid:", repeated_recording_ids)
```

```python
repeated_ids = data.loc[
    data["recording_msid"].duplicated(keep=False),
    "recording_msid"
]

example_id = repeated_ids.iloc[0]

data.loc[
    data["recording_msid"] == example_id,
    [
        "user_id",
        "user_name",
        "timestamp",
        "recording_msid",
        "track_metadata.track_name",
        "track_metadata.artist_name"
    ]
]
```

```python
missing_values = pd.DataFrame({
    "before cleaning": data[relevant_cols].isna().sum(),
    "after cleaning": clean[relevant_cols].isna().sum()
})

missing_values
```

```python
suspicious_counts = pd.Series({
    "Missing or invalid timestamp": clean["timestamp_clean"].isna().sum(),
    "Listening dates after September 29, 2026": (
        clean["timestamp_clean"].dt.date > pd.Timestamp("2026-09-29").date()
    ).sum(),
    "Blank artist names": clean["track_metadata.artist_name"].str.strip().eq("").sum(),
    "Blank track names": clean["track_metadata.track_name"].str.strip().eq("").sum()
})

suspicious_counts
```

```python
starting_rows = len(data)
final_rows = len(clean)

print("Starting rows:", starting_rows)
print("Final rows:", final_rows)
print("Difference:", final_rows - starting_rows)
```

```python
total_rows = len(clean)
known_release_rows = clean["track_metadata.release_name"].notna().sum()
missing_release_rows = clean["track_metadata.release_name"].isna().sum()

print("Total listening-event rows:", total_rows)
print("Rows with a known release name:", known_release_rows)
print("Rows with a missing release name:", missing_release_rows)
```

```python
# Record 1: normal record
record1 = data.iloc[0]

# Record 2: missing release name
record2 = data[
    data["track_metadata.release_name"].isna()
].iloc[0]

# Record 4: unusual Unicode artist name
record4 = data[
    data["track_metadata.artist_name"].str.contains("𝓒𝓾𝓻𝓻𝓮𝓷", na=False)
].iloc[0]

# Record 3: repeated recording_msid, excluding records already selected
record3 = data[
    data["recording_msid"].duplicated(keep=False)
    & ~data.index.isin([record1.name, record2.name, record4.name])
].iloc[0]

# Record 5: another normal record, excluding everything already selected
used_indices = [
    record1.name,
    record2.name,
    record3.name,
    record4.name
]

record5 = data[
    ~data.index.isin(used_indices)
].iloc[0]

check_indices = [
    record1.name,
    record2.name,
    record3.name,
    record4.name,
    record5.name
]

print("Selected row indices:", check_indices)
```

Expected results before checking
Record 1 (row 0): Keep the record. The timestamp should convert successfully to UTC, and the artist and track names should remain unchanged unless they contain leading or trailing whitespace.
Record 2 (row 176): Keep the record. The missing release name should remain missing because release information is not required for a valid listening event.
Record 3 (row 18): Keep the record even though its recording_msid is repeated. A repeated recording ID can represent a separate listening event.
Record 4 (row 4): Keep the record and preserve the unusual Unicode artist name because unusual formatting does not necessarily indicate incorrect data.
Record 5 (row 1): Keep the record. The timestamp should convert successfully, and already-clean artist and track text should remain unchanged.

```python
columns_to_check = [
    "timestamp",
    "timestamp_clean",
    "recording_msid",
    "track_metadata.artist_name",
    "track_metadata.track_name",
    "track_metadata.release_name"
]

clean.loc[check_indices, columns_to_check]
```
